Создайте такую функцию, которая преобразует входящий список в новый уменьшенный или такой же список. В итоговом списке должны остаться только простые числа (числа которые делятся только на 1 и на себя, например 7).

```python
def is_prime(n):
    div = 2

    if n == 1:
        return False
    elif n == 2:
        return True

    # Выполняется, пока не найден делитель n % div != 0
    while n % div != 0:
        div += 1

        if div == n:
            return True
        
    return False
```

```python
def get_prime_number_list(source_list: list):
    # Создать новый список с элементами-простыми числами из исходного списка
    return [element for element in source_list if is_prime(element)]
```

```python
numbers = [1, 10, 3, 263, 641, 9419, 1001, 24077, 22333, 22409, 30341, 12443132, 973476234, 723466453, 100001591]

get_prime_number_list(numbers)
```

Задача 4. Написание функции для создания словаря на основе числа Фибоначчи
Создайте функцию fibonacci_dict(n), которая принимает одно целое число n и возвращает словарь.

- Ключи словаря — это числа от 0 до n включительно.
- Значения словаря — это списки, содержащие все числа из последовательности Фибоначчи, которые меньше или равны ключу.

```python
# 0,1,1,2,3,5,8,13,21
# 0, 1 - 1
# 0, 1, 1 - 2
# 0, 1, 1, 2 - 3
# 0, 1, 1, 2, 3 - 4
# 0, 1, 1, 2, 3, 5 - 5

fib_cache = [0, 1]

def fibonacci_sequence(n: int):
    """
    Args:
        n (int): Fibbonaci number

    Returns:
        list: 
            [] - if bad n argument
            [0, 1, 1...<n_fibbonaci_number>] - list of fibbonaci numbers include n-number
    """
    if n < 0:
        return []

    if len(fib_cache) >= n + 1:
        return fib_cache[:n + 1]
    else:
        while len(fib_cache) < n + 1:
            fib_cache.append(fib_cache[-1] + fib_cache[-2])
        
        return fib_cache[:n + 1]

def fibonacci_dict(n: int):
    return {number: fibonacci_sequence(number) for number in range(n + 1)}

print(fibonacci_dict(5))
```
